```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from ucimlrepo import fetch_ucirepo

print("Libraries imported successfully")
```

```python
# Fetch AI4I 2020 Predictive Maintenance Dataset
ai4i = fetch_ucirepo(id=601)

# Get features and targets
X = ai4i.data.features
y = ai4i.data.targets

# Combine into one dataframe
df = pd.concat([X, y], axis=1)
print(f"Dataset shape: {df.shape}")
print(f"\nColumns:")
for col in df.columns.tolist():
    print(f"  → {col}")
```

```python
print("First 5 rows:")
df.head()
```

```python
print("Data Types:")
print(df.dtypes)
print(f"\nNull Values:")
print(df.isnull().sum())
print(f"\nTotal null values: {df.isnull().sum().sum()}")
```

```python
print("Failure Distribution:")
print(df['Machine failure'].value_counts())
print(f"\nFailure rate: {df['Machine failure'].mean()*100:.2f}%")
print(f"Normal rate: {(1-df['Machine failure'].mean())*100:.2f}%")

print("\nFailure Types Breakdown:")
failure_types = ['TWF', 'HDF', 'PWF', 'OSF', 'RNF']
for ft in failure_types:
    count = df[ft].sum()
    print(f"  {ft}: {count} cases")
```

```python
print("Sensor Statistics:")
sensors = [
    'Air temperature',
    'Process temperature', 
    'Rotational speed',
    'Torque',
    'Tool wear'
]
df[sensors].describe().round(2)
```

```python
print("Correlation with Machine Failure:")
numeric_cols = [
    'Air temperature',
    'Process temperature',
    'Rotational speed',
    'Torque',
    'Tool wear',
    'Machine failure'
]
correlation = df[numeric_cols].corr()['Machine failure'].sort_values(ascending=False)
print(correlation.round(4))
```

```python
sensor_cols = [
    'Air temperature',
    'Process temperature',
    'Rotational speed',
    'Torque',
    'Tool wear'
]

ranges = df[sensor_cols].agg(['min','max','mean','std']).T.round(2)
ranges.columns = ['Min', 'Max', 'Mean', 'Std']
print(ranges.to_string())
```
